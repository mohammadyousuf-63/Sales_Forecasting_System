```python
# ============================================================
# STAGE 6: FEATURE ENGINEERING
# Cell 1: Extract Year from Order Date
# ============================================================

# Extract the year from the Order Date
df["Year"] = df["Order Date"].dt.year

# Display the first 5 rows
df[["Order Date", "Year"]].head()
```

```python
# ============================================================
# STAGE 6: FEATURE ENGINEERING
# Cell 2: Extract Month from Order Date
# ============================================================

# Extract the month name from the Order Date
df["Month"] = df["Order Date"].dt.month_name()

# Display the first 5 rows
df[["Order Date", "Year", "Month"]].head()
```

```python
# ============================================================
# STAGE 6: FEATURE ENGINEERING
# Cell 3: Create Month Number
# ============================================================

# Extract the numeric month from Order Date
df["Month_Number"] = df["Order Date"].dt.month

# Display the first 5 rows
df[["Order Date", "Year", "Month", "Month_Number"]].head()
```

```python
# ============================================================
# STAGE 6: FEATURE ENGINEERING
# Cell 4: Create Time Index
# ============================================================

# Create a continuous time index based on Year and Month
df["Time_Index"] = (
    (df["Year"] - df["Year"].min()) * 12
    + (df["Month_Number"] - 1)
)

# Display the first 5 rows
df[["Order Date", "Year", "Month", "Month_Number", "Time_Index"]].head()
```

```python
# ============================================================
# STAGE 4: FEATURE ENGINEERING
# Cell 5: Extract Day Feature
# ============================================================

# Extract the day from Order Date
df["Day"] = df["Order Date"].dt.day

# Display the first 5 rows
print("Day feature created successfully.")

df[["Order Date", "Year", "Month", "Day"]].head()
```

```python
# ============================================================
# STAGE 6: FEATURE ENGINEERING
# Cell 6: Create Monthly Forecasting Dataset
# ============================================================

# Create a monthly dataset by summing Sales for each month
monthly_data = df.groupby(
    ["Year", "Month_Number"]
)["Sales"].sum().reset_index()

# Create a continuous time index for each month
monthly_data["Time_Index"] = (
    (monthly_data["Year"] - monthly_data["Year"].min()) * 12
    + (monthly_data["Month_Number"] - 1)
)

# Display the monthly forecasting dataset
monthly_data.head(10)
```

```python
# ============================================================
# STAGE 6: FEATURE ENGINEERING
# Cell 7: Check the engineered monthly dataset
# ============================================================

# Display the shape of the monthly dataset
print("Monthly Dataset Shape:", monthly_data.shape)

# Check the first and last month
print("First Month:", monthly_data.iloc[0]["Year"], "-", monthly_data.iloc[0]["Month_Number"])
print("Last Month:", monthly_data.iloc[-1]["Year"], "-", monthly_data.iloc[-1]["Month_Number"])

# Check for missing values
print("\nMissing Values:")
print(monthly_data.isnull().sum())

# Display the last 5 rows
print("\nLast 5 Months:")
monthly_data.tail()
```

```python
# ============================================================
# STAGE 6: FEATURE ENGINEERING
# Cell 8: Feature Engineering Summary
# ============================================================

print("Feature Engineering Completed Successfully!")
print("=" * 55)

print("\nFeatures Created in Original Dataset:")
print("1. Year")
print("2. Month")
print("3. Day")
print("4. Month_Number")
print("5. Time_Index")

print("\nMonthly Forecasting Dataset:")
print("Total Months:", len(monthly_data))
print("Start Year:", int(monthly_data["Year"].min()))
print("End Year:", int(monthly_data["Year"].max()))

print("\nFeatures Used for Monthly Forecasting:")
print("- Time_Index")
print("- Month_Number")
print("- Sales (Target Variable)")

print("\nFeature Engineering Stage Completed.")
```

```python
# ============================================================
# STAGE 7: MODEL BUILDING
# Cell 1: Prepare Features and Target
# ============================================================

# Features used by the model
X = monthly_data[["Time_Index", "Month_Number"]]

# Target variable: monthly sales
y = monthly_data["Sales"]

# Display the features
print("Features (X):")
print(X.head())

# Display the target
print("\nTarget (y):")
print(y.head())
```

```python
# ============================================================
# STAGE 7: MODEL BUILDING
# Cell 2: Chronological Train-Test Split
# ============================================================

# Use the first 36 months for training
X_train = X.iloc[:36]
y_train = y.iloc[:36]

# Use the last 12 months for testing
X_test = X.iloc[36:]
y_test = y.iloc[36:]

# Display the shapes of training and testing data
print("Training Features Shape:", X_train.shape)
print("Training Target Shape:", y_train.shape)

print("\nTesting Features Shape:", X_test.shape)
print("Testing Target Shape:", y_test.shape)
```

```python
# ============================================================
# STAGE 7: MODEL BUILDING
# Cell 3: Check Training and Testing Data
# ============================================================

# Display the last 5 training records
print("LAST 5 TRAINING RECORDS:")
print(monthly_data.iloc[:36].tail())

# Display the first 5 testing records
print("\nFIRST 5 TESTING RECORDS:")
print(monthly_data.iloc[36:].head())

# Display the training and testing time ranges
print("\nTraining Time Index:", X_train["Time_Index"].min(), "to", X_train["Time_Index"].max())
print("Testing Time Index:", X_test["Time_Index"].min(), "to", X_test["Time_Index"].max())
```

```python
# ============================================================
# STAGE 7: MODEL BUILDING
# Cell 4: Create Linear Regression Model
# ============================================================

# Import Linear Regression
from sklearn.linear_model import LinearRegression

# Create the Linear Regression model
model = LinearRegression()

# Display the model
print("Linear Regression model created successfully.")
print(model)
```

```python
# ============================================================
# STAGE 7: MODEL BUILDING
# Cell 5: Train the Linear Regression Model
# ============================================================

# Train the model using the training data
model.fit(X_train, y_train)

# Confirm that training is complete
print("Linear Regression model trained successfully.")
```

```python
# ============================================================
# STAGE 7: MODEL BUILDING
# Cell 6: Model Summary
# ============================================================

# Get the model coefficients
print("Model Coefficients:")
print("Time_Index:", model.coef_[0])
print("Month_Number:", model.coef_[1])

# Get the intercept
print("\nModel Intercept:")
print(model.intercept_)
```

```python
# ============================================================
# STAGE 8: PREDICTION
# Cell 1: Generate Predictions
# ============================================================

# Use the test features to predict monthly sales
y_pred = model.predict(X_test)

# Display the predicted sales
print("Predicted Sales:")
print(y_pred)
```

```python
# ============================================================
# STAGE 8: PREDICTION
# Cell 2: Compare Actual vs Predicted Sales
# ============================================================

# Create a comparison DataFrame
comparison = pd.DataFrame({
    "Actual Sales": y_test.values,
    "Predicted Sales": y_pred
})

# Display the comparison
comparison
```

```python
# ============================================================
# STAGE 8: PREDICTION
# Cell 3: Create Detailed Prediction Table
# ============================================================

# Create a detailed comparison table with year and month
prediction_data = monthly_data.iloc[36:][
    ["Year", "Month_Number"]
].copy()

# Add actual and predicted sales
prediction_data["Actual Sales"] = y_test.values
prediction_data["Predicted Sales"] = y_pred

# Reset the index
prediction_data = prediction_data.reset_index(drop=True)

# Display the detailed prediction table
prediction_data
```

```python
# ============================================================
# STAGE 8: PREDICTION
# Cell 4: Visualize Actual vs Predicted Sales
# ============================================================

# Create month labels for the 2017 test period
month_labels = prediction_data["Month_Number"].astype(str)

# Create the plot
plt.figure(figsize=(12, 6))

# Plot actual sales
plt.plot(
    month_labels,
    prediction_data["Actual Sales"],
    marker="o",
    label="Actual Sales"
)

# Plot predicted sales
plt.plot(
    month_labels,
    prediction_data["Predicted Sales"],
    marker="o",
    label="Predicted Sales"
)

# Add chart details
plt.title("Actual vs Predicted Monthly Sales - 2017")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()
plt.tight_layout()

plt.show()
```

```python
# ============================================================
# STAGE 8: PREDICTION
# Cell 5: Prediction Summary
# ============================================================

# Calculate total actual and predicted sales
total_actual_sales = y_test.sum()
total_predicted_sales = y_pred.sum()

# Calculate the difference
total_difference = total_actual_sales - total_predicted_sales

# Display the results
print("PREDICTION SUMMARY")
print("=" * 50)

print(f"Total Actual Sales:    {total_actual_sales:.2f}")
print(f"Total Predicted Sales: {total_predicted_sales:.2f}")
print(f"Difference:            {total_difference:.2f}")
```

```python
# ============================================================
# STAGE 9: MODEL EVALUATION
# Cell 1: Calculate Evaluation Metrics
# ============================================================

# Import evaluation metrics
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

# Calculate Mean Absolute Error
mae = mean_absolute_error(y_test, y_pred)

# Calculate Root Mean Squared Error
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

# Calculate R² Score
r2 = r2_score(y_test, y_pred)

# Display the evaluation results
print("MODEL EVALUATION")
print("=" * 50)

print(f"MAE:      {mae:.2f}")
print(f"RMSE:     {rmse:.2f}")
print(f"R² Score: {r2:.4f}")
```

```python
# ============================================================
# STAGE 9: MODEL EVALUATION
# Cell 2: Calculate Prediction Errors
# ============================================================

# Create a copy of the prediction data
evaluation_data = prediction_data.copy()

# Calculate the prediction error
# Positive value = model underestimated sales
# Negative value = model overestimated sales
evaluation_data["Error"] = (
    evaluation_data["Actual Sales"]
    - evaluation_data["Predicted Sales"]
)

# Calculate absolute error
evaluation_data["Absolute Error"] = (
    evaluation_data["Error"].abs()
)

# Display the evaluation table
evaluation_data
```

```python
# ============================================================
# STAGE 9: MODEL EVALUATION
# Cell 3: Visualize Prediction Errors
# ============================================================

# Create month labels
month_labels = evaluation_data["Month_Number"].astype(str)

# Create the error plot
plt.figure(figsize=(12, 6))

plt.bar(
    month_labels,
    evaluation_data["Error"]
)

# Add a zero reference line
plt.axhline(y=0, linestyle="--")

# Add chart details
plt.title("Prediction Error by Month - 2017")
plt.xlabel("Month")
plt.ylabel("Prediction Error")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# STAGE 9: MODEL EVALUATION
# Cell 4: Actual vs Predicted Evaluation Plot
# ============================================================

# Create the scatter plot
plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred
)

# Create a perfect prediction reference line
min_value = min(y_test.min(), y_pred.min())
max_value = max(y_test.max(), y_pred.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

# Add chart details
plt.title("Actual vs Predicted Sales")
plt.xlabel("Actual Sales")
plt.ylabel("Predicted Sales")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# STAGE 9: MODEL EVALUATION
# Cell 5: Final Evaluation Summary
# ============================================================

print("FINAL MODEL EVALUATION")
print("=" * 50)

print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R² Score: {r2:.4f}")

print("\nMODEL INTERPRETATION")
print("-" * 50)

print(f"The model explains {r2 * 100:.2f}% of the variation in test sales.")
print(f"Average prediction error is approximately {mae:.2f}.")
print(f"RMSE is {rmse:.2f}, indicating some larger prediction errors.")

print("\nConclusion:")
print("Linear Regression captures the overall sales trend,")
print("but it does not capture all month-to-month sales fluctuations.")
```

```python
# ============================================================
# STAGE 10: FUTURE SALES FORECASTING
# Cell 1: Identify Last Historical Month
# ============================================================

# Find the last year in the historical monthly dataset
last_year = int(monthly_data["Year"].max())

# Find the last month number in the historical monthly dataset
last_month = int(monthly_data.loc[
    monthly_data["Year"] == last_year, "Month_Number"
].max())

# Display the last historical period
print("Last Historical Year:", last_year)
print("Last Historical Month:", last_month)
```

```python
# ============================================================
# STAGE 10: FUTURE SALES FORECASTING
# Cell 2: Create Future Months
# ============================================================

# Number of future months to forecast
forecast_months = 12

# Create future time index values
future_time_index = np.arange(
    len(monthly_data),
    len(monthly_data) + forecast_months
)

# Create future year and month numbers
future_year = []
future_month = []

for i in range(1, forecast_months + 1):
    month_number = last_month + i
    year = last_year + (month_number - 1) // 12
    month = (month_number - 1) % 12 + 1
    
    future_year.append(year)
    future_month.append(month)

# Create the future forecasting DataFrame
future_data = pd.DataFrame({
    "Year": future_year,
    "Month_Number": future_month,
    "Time_Index": future_time_index
})

# Display the future months
print("Future months created successfully.")
print("\nFuture Forecast Period:")
print(future_data)
```

```python
# ============================================================
# STAGE 10: FUTURE SALES FORECASTING
# Cell 3: Prepare Future Features
# ============================================================

# Select the same features used when training the Random Forest model
X_future = future_data[["Time_Index", "Month_Number"]]

# Display the future features
print("Future forecasting features prepared successfully.")
print("\nFeatures used for prediction:")
print(X_future)
```

```python
# ============================================================
# STAGE 10: FUTURE SALES FORECASTING
# Cell 4: Generate Future Sales Predictions
# ============================================================

# Predict future sales using the trained Random Forest model
future_predictions = rf_model.predict(X_future)

# Add the predictions to the future dataset
future_data["Predicted_Sales"] = future_predictions

# Display the predictions
print("Future sales predictions generated successfully.")
print("\nFuture Sales Forecast:")
print(future_data)
```

```python
# ============================================================
# STAGE 11: FUTURE FORECAST VISUALIZATION
# Cell 1: Prepare Historical Sales Data
# ============================================================

# Create a copy of the monthly historical dataset
historical_data = monthly_data.copy()

# Create a date column for the historical months
historical_data["Date"] = pd.to_datetime(
    historical_data["Year"].astype(str)
    + "-"
    + historical_data["Month_Number"].astype(str)
    + "-01"
)

# Display the historical data prepared for visualization
print("Historical sales data prepared successfully.")
print(historical_data[["Date", "Sales"]].head())
```

```python
# ============================================================
# STAGE 11: FUTURE FORECAST VISUALIZATION
# Cell 2: Prepare Future Forecast Data
# ============================================================

# Create a copy of the future forecast dataset
forecast_data = future_data.copy()

# Create a date column for the future forecast months
forecast_data["Date"] = pd.to_datetime(
    forecast_data["Year"].astype(str)
    + "-"
    + forecast_data["Month_Number"].astype(str)
    + "-01"
)

# Display the future forecast data prepared for visualization
print("Future forecast data prepared successfully.")
print(forecast_data[["Date", "Predicted_Sales"]])
```

```python
# ============================================================
# STAGE 11: FUTURE FORECAST VISUALIZATION
# Cell 3: Create Combined Forecast Dataset
# ============================================================

# Prepare historical data for the combined dataset
historical_plot = historical_data[["Date", "Sales"]].copy()

# Rename the historical sales column
historical_plot.rename(
    columns={"Sales": "Actual_Sales"},
    inplace=True
)

# Prepare future forecast data for the combined dataset
forecast_plot = forecast_data[["Date", "Predicted_Sales"]].copy()

# Combine historical and future data
combined_forecast = pd.merge(
    historical_plot,
    forecast_plot,
    on="Date",
    how="outer"
)

# Sort the data by date
combined_forecast = combined_forecast.sort_values("Date").reset_index(drop=True)

# Display the combined dataset
print("Historical and future forecast data combined successfully.")
print("\nLast 15 rows:")
print(combined_forecast.tail(15))
```

```python
# ============================================================
# STAGE 11: FUTURE FORECAST VISUALIZATION
# Cell 4: Plot Historical Sales and Future Forecast
# ============================================================

# Create the plot
plt.figure(figsize=(14, 6))

# Plot historical actual sales
plt.plot(
    combined_forecast["Date"],
    combined_forecast["Actual_Sales"],
    label="Historical Sales"
)

# Plot future predicted sales
plt.plot(
    combined_forecast["Date"],
    combined_forecast["Predicted_Sales"],
    label="Future Forecast"
)

# Mark the beginning of the future forecast
plt.axvline(
    pd.Timestamp("2018-01-01"),
    linestyle="--",
    label="Forecast Start"
)

# Add chart title and labels
plt.title("Historical Monthly Sales and Future Sales Forecast")
plt.xlabel("Date")
plt.ylabel("Sales")

# Display legend and grid
plt.legend()
plt.grid(True)

# Display the chart
plt.show()
```

```python
# ============================================================
# STAGE 12: FINAL RESULTS & BUSINESS INTERPRETATION
# Cell 1: Future Forecast Summary
# ============================================================

# Calculate total predicted sales for 2018
total_forecast_sales = future_data["Predicted_Sales"].sum()

# Calculate average predicted monthly sales
average_forecast_sales = future_data["Predicted_Sales"].mean()

# Find the month with the highest predicted sales
highest_forecast_index = future_data["Predicted_Sales"].idxmax()
highest_forecast_month = future_data.loc[highest_forecast_index, "Month_Number"]
highest_forecast_sales = future_data.loc[
    highest_forecast_index, "Predicted_Sales"
]

# Find the month with the lowest predicted sales
lowest_forecast_index = future_data["Predicted_Sales"].idxmin()
lowest_forecast_month = future_data.loc[lowest_forecast_index, "Month_Number"]
lowest_forecast_sales = future_data.loc[
    lowest_forecast_index, "Predicted_Sales"
]

# Display the forecast summary
print("FUTURE SALES FORECAST SUMMARY")
print("=" * 50)

print(f"Forecast Period: January 2018 - December 2018")
print(f"Total Predicted Sales: {total_forecast_sales:.2f}")
print(f"Average Monthly Predicted Sales: {average_forecast_sales:.2f}")

print(f"\nHighest Predicted Sales:")
print(f"Month Number: {highest_forecast_month}")
print(f"Predicted Sales: {highest_forecast_sales:.2f}")

print(f"\nLowest Predicted Sales:")
print(f"Month Number: {lowest_forecast_month}")
print(f"Predicted Sales: {lowest_forecast_sales:.2f}")
```
